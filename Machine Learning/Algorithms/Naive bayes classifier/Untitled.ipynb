Task: News Category Prediction

You are given a dataset of news articles. Your task is to implement Naive Bayes from scratch and predict whether a new article belongs to:

🏏 Sports
💻 Technology
🏛️ Politics

```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Article": [
        "cricket team won match",
        "football player scored goal",
        "cricket player hit century",

        "new smartphone launched",
        "artificial intelligence improves software",
        "new laptop has powerful processor",

        "government announced new policy",
        "election campaign started",
        "minister announced new scheme"
    ],

    "Category": [
        "Sports",
        "Sports",
        "Sports",

        "Technology",
        "Technology",
        "Technology",

        "Politics",
        "Politics",
        "Politics"
    ]
}

df = pd.DataFrame(data)
df
```

```python
sport_count = np.sum(df["Category"] == "Sports")
tech_count = np.sum(df["Category"] == "Technology")
Politics=np.sum(df['Category']=="Politics")
#total
total_count = len(df)
p_sport = sport_count/total_count
p_tech = tech_count/total_count
p_politic=Politics/total_count
print(p_sport)
print(p_tech)
print(p_politic)
```

```python
def word_probablity(word,class_name):
  messages = df[df["Category"] ==class_name]["Article"]
  count = 0
  for message in messages:
    words = message.lower().split()
    if word in words:
      count +=1
    class_count = len(messages)
    probablity = (count+1)/(class_count +3)
  return probablity
```

```python
print("Calculated probability using functions")
print("Probabability of cricket in sports ",word_probablity('cricket','Sports'))
print("Probability of cricket in technology",word_probablity('cricket','Technology'))
```

```python
while True : 
    word = input('Enter a Word:') 
    if word == 'B': 
         print(' \nPrediction completed exiting the program...', '\n') 
         break
    sports_prob=p_sport*word_probablity(word, 'Sports') 
    tech_prob = p_tech*word_probablity(word, 'Technology ') 
    pol_prob = p_politic*word_probablity(word, 'Politics ') 
    print (' \nCombined probablities ') 
    print('='*20) 
    print("'Probability of given word in sports is :",sports_prob) 
    print('Probability of given word in technology is :',tech_prob) 
    print('Probability of given word in politics is :',pol_prob, '\n')
```
