2.Understanding Series

What is series?
  A Serie is like a single column in Excel-It's a list of values with labels(called index)

Index  Value
0 ---> Apple
1 ---> Banana
3 ---> Orange

```python
fruits=pd.Series(['Apple','Banana','Guava','Mango'])
print('Series of Fruits')
print(fruits)
print('\nType',type(fruits))

#here 0123 are Intrinsic index values which means default index
```

```python
#Creating a series with custom index values
veg=pd.Series([30,45,67,89],index=['Tomato','Potato','Lemon','Greens'])
print(veg)
print('Price of tomato',veg['Tomato'])
```

```python
#Series from a dictionary 
st_d=pd.Series({'syed':22,'baala':23,'Abinesh':20})
print(st_d)
```

Understanding Dataframe

Dataframe is like a excel spreadsheet that has rows and columns .It's a table with headers and data values.

Name   Age  Salary  Dept
------------------------
Alice  28   50000   IT
Bob    32   60000   HR
hari   25   45000   IT



```python
#Creating dataframe with dictionary
data={'Name':['Alice','Bob','Hari','John'],
      'Age':[28,32,25,21],
      'Salary':[50000,60000,45000,30000],
      'Dept':['IT','HR','IT','Finance']}
df=pd.DataFrame(data)
print('Employee data')
print(data)
print('Data frame')
print(df)
print(f'Shape:{df.shape}')
print(f'Type:{type(df)}')
```

```python
#methods to explore data
print('first two rows')
print(df.head(2))

print('----------------')
print('Last three rows')
print(df.tail(3))
print('----------------')

print('Data frame Info')
print(df.info())
print('---------------')

print('Statistical summary')
print(df.describe())
print('--------------------')
```

4. Accessing Data from Dataframe
**Different ways to get data**
-->'.loc[]' -Access by label(name)  #this is for user specified index
-->'.iloc[]'-Access by position(number) #similar to string indexing

```python
#Accessing columns,single columns,multiple columns
print(df.columns)
print(df['Name'])
print(df['Name'][:2])
print(df[['Name','Age','Salary']])
```

```python
#loc Access by rows and columns
print("Using .loc (by label):")
print("\nRow at index 1:")
print(df.iloc[1])
print("----------------------")
print("\nSpecific cell -Alice's salary:")
print(df.loc[0,'Salary'])
```

```python
df.loc[3,['Name','Age','Salary']]
```

```python
#Sorting
print("Employees sorted by age (youngest to oldest)")
print(df.sort_values('Age'))
print("----------------------------")
print("Employees sorted by salary from highest to lowest")
print(df.sort_values('Salary',ascending=False))
print('------------------------------------------')
```
