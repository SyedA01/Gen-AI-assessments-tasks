```python
#PART 4 — Pandas Foundations + NumPy Integration
data = {
    'Name': ['Arun', 'Priya', 'Karthik', 'Meena', 'Rahul', 'Divya'],
    'Age': [24, 29, 35, 27, 41, 31],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR', 'IT'],
    'Salary': [35000, 52000, 78000, 45000, 92000, 68000],
    'Experience': [2, 5, 10, 3, 15, 7]
}

df = pd.DataFrame(data)
df
#43. Display the first 3 rows.
print(df.head(3))
#Display the last 2 rows.
print(df.tail(2))
#44. Find the shape.
print(df.shape)
#45. Find the column names.
print(df.columns)
#46. Display information about the DataFrame.
print(df.info())
#47. Generate statistical information.
print(df.describe())
```

```python
#Selection
#48. Select Name.
print(df['Name'])
#49. Select Name and Salary.
print(df[['Name','Salary']])
#50. Select rows 1–3 using positional indexing.
print(df[:3])
#Select the third row using label-based indexing.
print(df.loc[2])
#Select the Name, Department, and Salary columns for rows 2–5.
print(df[['Name','Department','Salary']][1:])
```

```python
#NumPy + Pandas
#Convert the Salary column into a NumPy array.
sal=df['Salary'].to_numpy()
print(sal)
#Using NumPy, calculate the average salary.
print('Average salary',sal.mean())
#Using NumPy, find the employees' salary values above ₹60,000.
thres=sal>60000
print('Employee salary greater than 60,000',sal[thres])
#Using NumPy, calculate the standard deviation of salary.
print(f'Standard deviation of salary{sal.std():.2f}')
#Convert the resulting NumPy values back into a Pandas-friendly structure and create a new column indicating whether each employee earns above ₹60,000.
df['Earning above 60k']=thres
print(df)
```

```python
#filtering
#Find employees from IT.
print(df[df['Department']=='IT'])
#Find employees earning more than ₹60,000.
print(df[df['Salary']>60000])
#Find employees with more than 5 years of experience.
print(df[df['Experience']>5])
#Find HR employees earning more than ₹50,000.
print(df[(df['Department']=='HR') & (df['Salary']>50000)])
#Find employees aged above 30 AND earning above ₹60,000.
print(df[(df['Age']>30) & (df['Salary']>60000)])
#Find employees from IT OR HR.
print(df[(df['Department']=='HR') | (df['Department']=='IT')])
#Find employees NOT from Finance.
print(df[~(df['Department']=='Finance')])
#Find employees whose salary is between ₹40,000 and ₹80,000.
print(df[(df['Salary']>39000) & (df['Salary']<81000)])
```

```python
#pandas-based classification
#Salary_Level
#Salary < 40000       → Low
#40000–70000          → Medium
#>70000               → High
def salary_level(salary):
    if salary>70000:
        return "High"
    elif salary>40000 and salary<70000:
        return "Medium"
    else:
        return "Low"
#Experience_Level

'''<3       → Junior
3–7      → Mid
>7       → Senior'''
def exp_level(experience):
    if experience>7:
        return "Senior"
    elif experience>=3 and experience<=7:
        return "Mid"
    else:
        return "Junior"
df['Salary_level']=df['Salary'].apply(salary_level)
df['Experience_level']=df["Experience"].apply(exp_level)
df
```

```python
#Create a Boolean column indicating employees eligible for promotion:
df['Bonus_eligible']=(df['Experience']>5) & (df['Salary']>60000)
df
```

```python
#PART 6 — Pandas Calculations + GroupBy
sales = pd.DataFrame({
    'Employee': ['Arun', 'Priya', 'Karthik', 'Meena', 'Rahul',
                 'Divya', 'Sanjay', 'Anitha'],
    'Department': ['IT', 'IT', 'HR', 'HR', 'Finance',
                   'Finance', 'IT', 'HR'],
    'Sales': [45000, 62000, 38000, 55000, 72000,
              68000, 51000, 47000],
    'Units': [12, 18, 9, 15, 21, 19, 14, 11]
})
#Calculate total sales.
print(f"Total Sales {sales['Sales'].sum():.2f}")
#Calculate average sales.
print(f"Average sales {sales['Sales'].mean():.1f}")
#Calculate total units.
print(f"Total units {sales['Units'].sum()}")
#Create a Revenue_Per_Unit column.
sales['Revenue_per_unit']=(sales['Sales']/sales['Units']).round().astype(int)
print(sales)
#Find the employee with the highest sales.
high=sales.loc[sales['Sales'].idxmax()]
print(high)
#Find the employee with the lowest sales.
low=sales.loc[sales['Sales'].idxmin()]
print(low)
```

```python
#Grouping
#Calculate total sales by department wise
total_sales=sales.groupby('Department')['Sales'].sum()
print("Department wise sales total\n",total_sales)
#Calculate average sales by department.
Avg_sales=sales.groupby('Department')['Sales'].mean()
print("Department wise sales average\n",Avg_sales)
#Calculate total units by department.
Total_units=sales.groupby('Department')['Units'].sum()
print("Departemnt wise units total\n",Total_units)
#Find the maximum sales in each department.
Maximum_sales=sales.groupby('Department')['Sales'].max()
print("Department wise maximum sales\n",Maximum_sales)
#Count employees in each department.
Count=sales.groupby('Department')['Employee'].count()
print("Department wise employees count\n",Count)
```

```python
#For each department calculate:
'''Total Sales
Average Sales
Maximum Sales
Minimum Sales
Total Units'''
aggregate=sales.groupby('Department')['Sales'].agg(['sum','mean','max','min'])
Total_units=sales.groupby('Department')['Units'].sum()
print("Departemnt wise units total\n",Total_units)
print(aggregate)
```
